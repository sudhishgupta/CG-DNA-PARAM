```python
import numpy as np
from scipy.signal import savgol_filter
```

```python
writhe = np.loadtxt('neg_wr22/writhe.ser')
tw = np.load('neg_wr22/workspace.pkl')['tw']
tw_mean = np.mean(tw, axis=1)
frames = [int(i) for i in range(1,len(writhe)+1)]
```
